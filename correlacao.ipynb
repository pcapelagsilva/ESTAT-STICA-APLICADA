**CORRELAÇÃO POSITIVA**

O resultado será próximo de *1* porque os genes aumentam juntos

```python
import pandas as pd

dados = {
    "Gene_A": [10, 20, 30, 40, 50],
    "Gene_B": [12, 22, 29, 41, 55]
}

df = pd.DataFrame(dados)

correlacao = df["Gene_A"].corr(df["Gene_B"])

print("Correlação: ", correlacao)
```

**CORRELAÇÃO NEGATIVA**

O resultado será próximo de *-1*

```python
import pandas as pd

dados = {
    "Gene_A": [10, 20, 30, 40, 50],
    "Gene_B": [50, 40, 30, 20, 10]
}

df = pd.DataFrame(dados)

correlacao = df["Gene_A"].corr(df["Gene_B"])

print("Correlação: ", correlacao)
```

**CALCULANDO O VALOR *r* E O *P-VALUE***

r --> Próximo de *1*

*p-value* --> Muito pequeno

Resultado: Forte correlação positiva significativa.

```python
from scipy.stats import pearsonr

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

r, p = pearsonr(x, y)

print("Correlação: ", r)
print("p-value: ", p)
```
